
-   Code the solution to your assigned problem in Python (code chunk). Note: each problem can be solved more simply if you use an abstract data type that is suitable for that problem. Using that try to find the best time and space complexity solution!


````python
from typing import List

def move_zeros_to_end(nums: List[int]) -> List[int]:
    position = 0

    for i in range(len(nums)):
        if nums[i] != 0:
            nums[position] = nums[i]
            position += 1

    for i in range(position, len(nums)):
        nums[i] = 0

    return nums
````


-   Explain why your solution works


````python
# My solution works by moving all the non-zero numbers to the front of the original list, then replacing the remaining positions with zeros. It modifies the list directly without creating a new list.

# **Position variable**
# The position variable records where the next non-zero integer should be placed.
# When position = 0, it means the first non-zero integer should go to index 0 of list nums.

# **First for loop**
# This loop runs through every integer in list nums.
# If nums[i] is not zero, then nums[i] is copied into nums[position].
# Then position increases by 1, meaning the next non-zero integer should be placed at the next index.

# As the for loop goes throught the full list, there will be old duplicated values at the end of the list.

# **Second for loop**
# This loop changes everything from index[position] to the end of the list into zeros.
````

-   Explain the problem’s time and space complexity


````python
# The time complexity is O(n), where n is the length of the list nums.

# The first for loop goes through every element in the list once to move all non-zero numbers to the front. The second for loop fills the remaining positions with zeros. Even though there are two loops, they are not nested, so the total time complexity is still O(n).

# The space complexity is O(1) because the solution modifies only one value at a time on the original list directly.
````
