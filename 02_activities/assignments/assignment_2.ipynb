#### Starter Code for Question 2

```python
# Definition for a binary tree node.
# class TreeNode(object):
#     def __init__(self, val = 0, left = None, right = None):
#         self.val = val
#         self.left = left
#         self.right = right

def bt_path(root: TreeNode) -> List[List[int]]:
    result = []

    def dfs(node, path):
        if not node:
            return

        path.append(node.val)

        if not node.left and not node.right:
            result.append(path.copy())
        else:
            dfs(node.left, path)
            dfs(node.right, path)

        path.pop()

    dfs(root, [])
    return result
```

```python
from typing import List

def first_duplicate(nums: List[int]) -> int:
    seen = set()

    for num in nums:
        if num in seen:
            return num
        seen.add(num)

    return -1
```


-   Explain why their solution works in your own words.


Their solution uses a set to keep track of the numbers they have already seen while going through the list.

An empty set 'seen' is created
A for loop goes through each number in the list from left to right, recording each number into the set.
If the number is already in the set, that means the number has appeared before, so it is the first duplicate encountered and is returned immediately.
If not, the number is added to the set and the loop continues.

After going through all the integers and no duplicate is found, the function returns '-1'.


-   Explain the problem’s time and space complexity in your own words.


The for loop goes through every character in the string once. Each character is processed one time only, so the total time complexity is O(n), where n is the length of the input.

The space complexity is O(n) because the function uses a set to store numbers that have already appeared. In the worst case, there can be no duplicates, and therefore the set would be the same size as the input, where n is the length of the input.
